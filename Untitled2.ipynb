```python
# estrurura de decisões

# escrever um programa que recebe o nome e a idade do usuário e verifica se o usuário é criança, adolescente e adulto

# <= 17 anos é menor de idade
# > 17 e <= 59 anos é adulto
# > 59 anos é adulto

# entrada
idade = int (input ("digite sua idade: "))

# processamento / saída

if idade <= 17:
  print (f" você tem {idade} anos de idade, você é menor de idade")
elif idade <= 59:
  print (f" você tem {idade} anos de idade, você é adulto")
else:
  print (f" você tem {idade} anos de idade, você é idoso")
```

```python
# calculo de porcentagem

# Crie um programa que leia o valor de uma compra e aplique o desconto

# <= R$100 nenhum desconto
# > R$100 e <= R$500 aplica 10% de desconto
# > R%500 aplica 20% de desconto

# entrada
valor_da_compra = int (input("valor"))

# processamento / saída
if valor_da_compra <=100:
  print (f"Valor final da compra: {valor_da_compra}")
elif valor_da_compra <= 500:
  print (f"Valor final da compra: {valor_da_compra * 0.9}")
else:
  print (f"Valor final da compra: {valor_da_compra * 0.8}")
```

```python
# encrever um programa que classifica os tipos de triângulo e valida se é um triângulo.

# validar se os lados digitados podem formar um triângulo
# lado1 + lado2> lado3

# entrada

lado1 = int (input ("digite o lado 1: "))
lado2 = int (input ("digite o lado 2: "))
lado3 = int (input ("digite o lado 3: "))

# verifica se é um triangulo válido

if lado1 + lado2 > lado3:
  print ("é um triangulo")
else:
   print ("não é um triangulo")
```

```python
# em um jogo, um personagem possui uma varíavel chamada energia. o programa deve:
# se a energia for maior que 70 - forte, se a energia estiver entre 30 e 70 - cansado, se a energia estiver menor que 30 - fraco.

# entrada

energia = int (input("digite a energia do seu personagem "))

#processamento / saída

if energia < 1:
  print (f"tá morto.")
elif energia <= 30:
  print (f"estado crítico.")
elif energia <= 70:
  print (f"normal.")
else:
  print (f" está saudável")
```

```python
# criar um programa que calcule a media do aluno
# entrada

nota_1_bimestre = float(input("digite a nota do 1 bimestre "))
nota_2_bimestre = float(input("digite a nota do 2 bimestre "))
nota_3_bimestre = float(input("digite a nota do 3 bimestre "))
nota_4_bimestre = float(input("digite a nota do 4 bimestre "))

# processamento
media = (nota_1_bimestre + nota_2_bimestre + nota_3_bimestre + nota_4_bimestre) / 4
if media >= 7:
  print (f"aprovado")

elif media >= 5:
  print (f"recuperação")

else:
  print (f"reprovado")
# saída

print(f"media: {media}")
```

```python
# Crie um programa que leia o salário de um funcionário e sua pontuação de desempenho

# Pontuação menor que 5: sem bônus
# Pontuação de 5 a 7: bônus de 5% do salário
# Pontuação de 8 a 9: bônus de 10% do salário
# Pontuação igual a 10: bônus de 15% do salário

# entrada

salario = float(input("digite o salário "))
pontuacao = int(input("digite a pontuação "))

# processamento

if pontuacao < 5:
  bonus = 0
elif pontuacao <= 7:
  bonus = salario * 0.05

elif pontuacao <= 9:
  bonus = salario * 0.1
else:
  bonus = salario * 0.15

# saída

print(f"salário: {salario}")
print(f"bonus: {bonus}")
```

```python
# Solicite ao usuário um número inteiro positivo.

#Utilizando um laço de repetição, calcule e exiba a soma de
#todos os números de 1 até o número informado.


# entrada
numero = int(input("Digite um número inteiro positivo: "))

soma = 0
for i in range(1, numero + 1):
    soma += i

print("Resultado:", soma)
```

```python
# Solicita um número inteiro

# entrada

numero = int(input("Digite um número: "))
for i in range(1, 11):
    print(f"{numero} x {i} = {numero * i}")
```

```python
# Solicite ao usuário um número inteiro positivo e calcule seu fatorial utilizando um laço de repetição.

# entrada

numero = int(input("Digite um número inteiro positivo: "))

fatorial = 1
for i in range(1, numero + 1):
    fatorial *= i

print("Resultado:", fatorial)
```

```python
# Um caixa eletrônico possui notas de:

# R$ 100
# R$ 50
# R$ 20
# R$ 10
# R$ 5
# R$ 2
# R$ 1

# Solicite ao usuário o valor do saque e informe a menor
# quantidade possível de notas para realizar o pagamento.

# Exemplo:

# Digite o valor do saque: 287

# Saída esperada:

# 2 nota(s) de R$100
# 1 nota(s) de R$50
# 1 nota(s) de R$20
# 1 nota(s) de R$10
# 1 nota(s) de R$5
# 1 nota(s) de R$2

# entrada

valor = int(input("Digite o valor do saque: "))

notas = [100, 50, 20, 10, 5, 2, 1]
for nota in notas:

    quantidade = valor // nota
    valor %= nota
    if quantidade > 0:
        print(f"{quantidade} nota(s) de R${nota}")
# processamento


```

```python
# um programa que determina s eum número é par ou impar

# entrada
numero = int(input("Digite um número: "))

# processamento
if numero % 2 == 0:
    print("O número é par.")
else:
    print("O número é ímpar.")
```

```python
# um programa que determina qual de 2 números é maior

# entrada
numero1 = int(input("Digite o primeiro número: "))
numero2 = int(input("Digite o segundo número: "))

# processamento
if numero1 > numero2:
    print("O primeiro número é maior.")
elif numero2 > numero1:
    print("O segundo número é maior.")
else:
    print("Os números são iguais.")
```

```python
# classifique a idade do usuário da seguinte maneira
# Criança 0-12
# Adolescente 13-17
# Adulto 18-59
# Idoso 60+

# entrada
idade = int(input("Digite a idade do usuário: "))

# processamento
if idade >= 0 and idade <= 12:
    print("Criança")
elif idade >= 13 and idade <= 17:
    print("Adolescente")
elif idade >= 18 and idade <= 59:
    print("Adulto")
else:
    print("Idoso")
```

```python
# Peça dois números e uma operação (+, -, *, /). Use if/elif para calcular o resultado.

# entrada
numero1 = float(input("Digite o primeiro número: "))
numero2 = float(input("Digite o segundo número: "))
operacao = input("Digite a operação (+, -, *, /): ")

# processamento
if operacao == "+":
    resultado = numero1 + numero2
elif operacao == "-":
    resultado = numero1 - numero2
elif operacao == "*":
    resultado = numero1 * numero2
elif operacao == "/":
    resultado = numero1 / numero2

# saída
print("O resultado da operação é:", resultado)
```

```python
# Peça um número de 1 a 7 e mostre o dia correspondente (ex: 1 = domingo).

# entrada

número =int(input("Digite um número de 1 a 7: "))

# processamento
if número == 1:
    print("Domingo")
elif número == 2:
    print("Segunda-feira")
elif número == 3:
    print("Terça-feira")
elif número == 4:
    print("Quarta-feira")

elif número == 5:
    print("Quinta-feira")
elif número == 6:
    print("Sexta-feira")
elif número == 7:
    print("Sábado")
else:
    print("Número inválido")
```

```python
# Peça o valor de uma compra: 	≥ 100: 10% de desconto 	≥ 50 e < 100: 5% de desconto  < 50: sem desconto

# entrada
compra = float(input("Digite o valor da compra: "))

# processamento
if compra >= 100:
    desconto = compra * 0.1
elif compra >= 50:
    desconto = compra * 0.05
else:
    desconto = 0

# saída
print("O valor do desconto é:", desconto)
print("O valor a ser pago é:", compra - desconto)
```

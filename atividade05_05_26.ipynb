```python
# escrever um programa que receba um numero n e imprima uma pirâmide perfeitamente centralizada

# entrada

n = int(input("Digite um número: "))

# processamento

for i in range(1, n + 1):
    print(" " * ((n - i) // 3) + "*" * i)

#
```

```python
# escrever um programa que o usuário digita 04 números e o programa imprimi o maior

# entrada

numero1 = int(input("Digite o primeiro número: "))
numero2 = int(input("Digite o segundo número: "))
numero3 = int(input("Digite o terceiro número: "))
numero4 = int(input("Digite o quarto número: "))

# processamento

maior = numero1 # Inicializa 'maior' com o primeiro número

if numero2 > maior:
    maior = numero2
if numero3 > maior:
    maior = numero3
if numero4 > maior:
    maior = numero4

# saída

print(f"O maior número é: {maior}")
```

```python
# escrever um programa que recebe as notas dos 04 bimestre e imprimi  a média com o resultado de aprovado, recuperação ou reprovado

# regras
# utilizar while
# nota igual ou maior que 7 aprovado
# nota igual ou maior que 5 recuperação
# nota menor que 5 reprovado

# entrada

nota = 0
bimestre = 1
resultado = 0

while bimestre <= 4:
      nota = float(input(f"Digite a nota do {bimestre}º bimestre: "))
      bimestre +=1
      resultado = resultado + nota
if (resultado /4) >= 7:
    print("Aprovado")
elif (resultado /4) >= 5:
    print("Recuperação")
else:
    print("Reprovado")
```

```python
# projeto de caixa eletrônico
# você foi contratado para desenvolver um sistema simples de caixa eletrônico
# o sistema deverá funcionar através de um menu interativo utilizando "while"
# o programa deve iniciar com um saldo inicial de R$1000,00 e permitir que o usuário escolha diferentes operações até decidir sair do sistema
# opção 1 sacar
# opção 2 depositar
# opção 3 ver saldo
# opção 4 sair do banco

# entrada
saldo = 1000.00

# processamento e saída
while True:
    print("\n--- MENU DO CAIXA ELETRÔNICO ---")
    print("1. Sacar")
    print("2. Depositar")
    print("3. Ver Saldo")
    print("4. Sair do Banco")

    try:
        opcao = int(input("Escolha uma opção: "))

        if opcao == 1: # Sacar
            valor_saque = float(input("Digite o valor a sacar: R$"))
            if valor_saque > saldo:
                print("Saldo insuficiente!")
            elif valor_saque <= 0:
                print("Valor de saque inválido. Digite um valor positivo.")
            else:
                saldo -= valor_saque
                print(f"Saque de R${valor_saque:.2f} realizado com sucesso.")
                print(f"Seu novo saldo é: R${saldo:.2f}")
        elif opcao == 2: # Depositar
            valor_deposito = float(input("Digite o valor a depositar: R$"))
            if valor_deposito <= 0:
                print("Valor de depósito inválido. Digite um valor positivo.")
            else:
                saldo += valor_deposito
                print(f"Depósito de R${valor_deposito:.2f} realizado com sucesso.")
                print(f"Seu novo saldo é: R${saldo:.2f}")
        elif opcao == 3: # Ver Saldo
            print(f"Seu saldo atual é: R${saldo:.2f}")
        elif opcao == 4: # Sair
            print("Obrigado por usar nosso caixa eletrônico. Volte sempre!")
            break # Sai do loop while
        else:
            print("Opção inválida. Por favor, escolha uma opção de 1 a 4.")
    except ValueError:
        print("Entrada inválida. Por favor, digite um número para a opção.")
```

```python
def somar(num1, num2):
    return num1 + num2

def subtrair(num1, num2):
    return num1 - num2

def multiplicar(num1, num2):
    return num1 * num2

def dividir(num1, num2):
    if num2 == 0:
        return "Erro: Divisão por zero!"
    return num1 / num2

while True:
    # calculadora com as 5 opções

    # menu
    print("\n--- Calculadora --- ")
    print("opção 1 - Somar")
    print("opção 2 - subtrair")
    print("opção 3 - Multiplicar")
    print("opção 4 - Divisão")
    print("opção 5 - Sair")
    # entrada
    escolha = input("Escolha uma opção (1-5): ")

    if escolha == '5':
        print("Saindo da calculadora. Até mais!")
        break

    if escolha in ('1', '2', '3', '4'):
        try:
            num1 = float(input("Digite o primeiro número: "))
            num2 = float(input("Digite o segundo número: "))
        except ValueError:
            print("Entrada inválida. Por favor, digite números.")
            continue

        if escolha == '1':
            print(f"Resultado: {num1} + {num2} = {somar(num1, num2)}")
        elif escolha == '2':
            print(f"Resultado: {num1} - {num2} = {subtrair(num1, num2)}")
        elif escolha == '3':
            print(f"Resultado: {num1} * {num2} = {multiplicar(num1, num2)}")
        elif escolha == '4':
            resultado_divisao = dividir(num1, num2)
            print(f"Resultado: {num1} / {num2} = {resultado_divisao}")
    else:
        print("Opção inválida. Por favor, escolha um número entre 1 e 5.")
```

```python
# Jogo Inteligente

def par_impar(num):
    if num % 2 == 0:
        return "Par"
    else:
        return "Ímpar"

def verificar_positivo(num):
    if num > 0:
        return "Positivo"
    elif num < 0:
        return "Negativo"
    else:
        return "Zero"

def maior_de_dois(num1, num2):
    if num1 > num2:
        return num1
    elif num2 > num1:
        return num2
    else:
        return "Ambos são iguais"

while True:
    # menu
    print("\n--- Jogo Inteligente ---")
    print("1 - Par / Ímpar")
    print("2 - Positivo / Negativo")
    print("3 - Maior de 02 números")
    print("4 - Sair")

    # entrada
    escolha_jogo = input("Escolha uma opção (1-4): ")

    if escolha_jogo == '4':
        print("Saindo do Jogo Inteligente. Até mais!")
        break

    if escolha_jogo in ('1', '2', '3'):
        if escolha_jogo == '1' or escolha_jogo == '2':
            try:
                numero_unico = float(input("Digite um número: "))
            except ValueError:
                print("Entrada inválida. Por favor, digite um número.")
                continue

            if escolha_jogo == '1':
                print(f"O número {numero_unico} é {par_impar(numero_unico)}.")
            elif escolha_jogo == '2':
                print(f"O número {numero_unico} é {verificar_positivo(numero_unico)}.")
        elif escolha_jogo == '3':
            try:
                num1_jogo = float(input("Digite o primeiro número: "))
                num2_jogo = float(input("Digite o segundo número: "))
            except ValueError:
                print("Entrada inválida. Por favor, digite números.")
                continue

            resultado_maior = maior_de_dois(num1_jogo, num2_jogo)
            if isinstance(resultado_maior, str):
                print(resultado_maior)
            else:
                print(f"O maior número entre {num1_jogo} e {num2_jogo} é {resultado_maior}.")
    else:
        print("Opção inválida. Por favor, escolha um número entre 1 e 4.")
```

```python
# Escreve um programa que verifica a senha, o usuário terá 3 tentativas caso erre as 03 imprimir mensagem de "acesso negado e tente novamente mais tarde!" e encerrar o programa.

senha_correta = "minhasenha123"
 tentativas = 0
 max_tentativas = 3

while tentativas < max_tentativas:
    # entrada
    senha_digitada = input(f"Tentativa {tentativas + 1}/{max_tentativas}: Digite a senha: ")

    if senha_digitada == senha_correta:
        print("Acesso concedido!")
        break
    else:
        print("Senha incorreta.")
        tentativas += 1

if tentativas == max_tentativas:
    print("Acesso negado e tente novamente mais tarde!")
```
